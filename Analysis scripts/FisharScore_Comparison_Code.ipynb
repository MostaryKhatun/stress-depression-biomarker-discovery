```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


# ==========================================================
# Directory
# ==========================================================

selection_dir = "/content/drive/MyDrive/p5_stress_depression_journal/process_datasets/selection"

score_dir = "/content/drive/MyDrive/p5_stress_depression_journal/process_datasets/score"

output_dir = "/content/drive/MyDrive/p5_stress_depression_journal/process_datasets/Comparison"

os.makedirs(output_dir, exist_ok=True)



# ==========================================================
# Select Dataset
# ==========================================================

dataset_name = input(
    "Enter dataset name (Example: GSE63878_matrix_Transpose_labeled): "
)


# Fisher Score file

score_file = os.path.join(
    score_dir,
    f"{dataset_name}_Fisher_Scores.csv"
)


fisher_df = pd.read_csv(score_file)



# Sort Fisher Score

fisher_df = fisher_df.sort_values(
    by="Fisher Score",
    ascending=False
)



# ==========================================================
# Compare Top 500,800,1000
# ==========================================================


top_values = [500,800,1000]


comparison = []



for top_n in top_values:


    selected_features = fisher_df.head(top_n)



    scores = selected_features["Fisher Score"]



    comparison.append({

        "Feature Selection": f"Top {top_n}",

        "Number of Features": top_n,

        "Mean Fisher Score": scores.mean(),

        "Median Fisher Score": scores.median(),

        "Maximum Fisher Score": scores.max(),

        "Minimum Fisher Score": scores.min(),

        "Total Fisher Score": scores.sum(),

        "Std Fisher Score": scores.std()

    })




comparison_df = pd.DataFrame(comparison)



# ==========================================================
# Display Table
# ==========================================================


print("\nFisher Score Feature Selection Comparison\n")

display(comparison_df)



# ==========================================================
# Save Table
# ==========================================================


table_file = os.path.join(
    output_dir,
    f"{dataset_name}_Top500_Top800_Top1000_Comparison.csv"
)


comparison_df.to_csv(
    table_file,
    index=False
)


print("\nComparison table saved:")
print(table_file)



# ==========================================================
# Plot Comparison
# ==========================================================


plt.figure(
    figsize=(10,6)
)


plt.bar(
    comparison_df["Feature Selection"],
    comparison_df["Mean Fisher Score"]
)


plt.xlabel(
    "Selected Features"
)


plt.ylabel(
    "Mean Fisher Score"
)


plt.title(
    f"{dataset_name}\nTop 500 vs Top 800 vs Top 1000 Feature Comparison"
)


plt.grid(
    axis="y"
)


plot_file = os.path.join(
    output_dir,
    f"{dataset_name}_Feature_Selection_Comparison.png"
)



plt.savefig(
    plot_file,
    dpi=1000,
    bbox_inches="tight"
)



plt.show()



print("\nPlot saved:")
print(plot_file)
```
